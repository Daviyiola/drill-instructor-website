# DIRI 3.1 calibration

## tl;dr
- DIRI 90 is achievable: 90% accuracy, 300 questions, and 15 active days scored 91.8.
- The current model can also award 90.4 at 70% accuracy when consistency and coverage are perfect.
- A 500-question one-day cram scored Ready (86.7), while equally strong work ending 60 days ago still scored 90.8.
- Focused Math rose from 81.3 to 90.9 when unrelated Science activity was added, without changing Math evidence.


## Context & Methods

This notebook reads deterministic outputs produced by the exact server `readiness()` function. Synthetic histories use the live ACT/SAT catalogs and a fixed clock so results are repeatable. These are calibration cases, not real-world predictive validation.

### Key Assumptions
- DIRI uses the trailing 90 days.
- Overall estimates require 100 answered questions; focused estimates require 40.
- Current weights are Performance 40%, Consistency 30%, Coverage 30%.
- Unless a scenario says otherwise, pacing is 60 seconds used against 60 seconds allocated.


````python
import json
from pathlib import Path
results = json.loads(Path('simulation-results.json').read_text(encoding='utf-8'))
results['formulaVersion'], results['checks']
````

## Data

Selected scenario output from the executed simulation harness:


````python
selected = {row['id']: row for row in results['scenarios']}
for key in ['minimum-balanced', 'balanced-300', 'accuracy-70-max', 'fresh-cram', 'stale-60', 'one-subject-broad', 'timer-generous', 'focused-math-plus-other-days']:
    row = selected[key]
    print(key, row['result']['score'], row['result'].get('pillars'))
````

## Takeaways

1. DIRI is monotonic in accuracy, but current weights allow activity and breadth to rescue weak performance too aggressively.
2. Recency and active-day breakpoints do not prevent cramming or stale high scores from remaining Ready.
3. Coverage must be computed within each subject before aggregation; current global numerators permit cross-subsidization.
4. Pacing should use canonical exam allocations, not a timer selected for an individual drill.
5. Native offline mode should stop recomputing a different formula and instead show the cached canonical DIRI.
